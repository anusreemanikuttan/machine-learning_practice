**Marker Properties**

* ms -> marker size

* mfc -> marker face color

* mec -> marker edge color

```python
x = [2,3,1,4]
y = [6,4,8,3]
plt.plot(x,y, marker='*', ms=15, mfc="r", mec='g')
plt.show()
```

**Line Properties**

* linestyle -> dotted , solid ..

* linewidth

* color

```python
x = [4,1,3,8]
y = [5,3,7,2]
plt.plot(x,y,marker="*", linestyle=":", linewidth=2, color="r",ms = 10, mfc="m", mec="g")
plt.show()
```

```python
x1 = [3,2,1,7,4]
y1 = [5,1,9,4,3]

x2 = [4,8,5,1,2]
y2 = [2,5,6,8,4]

plt.plot(x1,y1, marker="*", linestyle="-.",ms=5, mfc="r", mec="g")
plt.plot(x2,y2, marker="o", linestyle=":",ms=5, mfc="m", mec="b")
plt.show()
```

* Title for axis -> label

* Title for graph -> title

```python
# Height

height = [152,165,178,183,174]
weight = [52,65,78,80,60]
plt.plot(height,weight,marker="*",color="g",mfc="r",mec="r")
plt.xlabel("Height", color="g")
plt.ylabel("Weight", color="g")
plt.title('Height-Weight Graph', color="m", fontsize=10)
plt.grid()     # grid format
plt.show()
```

```python
# Draw a velocity - displacement graph

v = [10,20,30,40,50]
d = [5,10,15,25,35]
plt.plot(v,d,marker="*",color="r",mfc="k",mec="k")
plt.xlabel("Velocity", color="g")
plt.ylabel("Displacement", color="g")
plt.title('Velocity-Displacement Graph', color="b", fontsize=10)
plt.grid()     # grid format
plt.show()
```

# **Scatter Plot**

# **Sub Plot**

```python
x1 = [4,3,5,2,1]
y1 = [5,3,7,9,4]
plt.subplot(1,2,1)
plt.plot(x1,y1, color="r")

x2 = [6,3,9,5,7]
y2 = [1,6,2,8,4]
plt.subplot(1,2,2)
plt.plot(x2,y2, color="g")
plt.suptitle("Graph1 - Graph2")
plt.show()
```

# **Bar Chart**

```python
import matplotlib.pyplot as plt

x = [5,6,2,3,1]
y = [6,2,3,9,4]
plt.bar(x,y)
plt.show()
```

```python
country = ["india", "us", "uk", "sweden", "swiss"]
count = [5,3,14,22,10]
plt.bar(country, count)
plt.show()
```

```python
# horizontal
country = ["india", "us", "uk", "sweden", "swiss"]
count = [5,3,14,22,10]
plt.barh(country, count)
plt.show()
```

# **Histogram**

# **Pie Chart**

```python
x = [20,30,25,10]
label = ["apple", "mango", "orange", "melon"]
plt.pie(x, labels=label,autopct='%1.1f%%', explode=[0,0.1,0,0])
plt.show()
```
